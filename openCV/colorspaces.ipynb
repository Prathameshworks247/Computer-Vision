```python
#rgb->bgr

#grascale - efficient for processing
#simplification, saves space, faster processing

#HSV - hue saturation value
#hue - color type
#saturation - intensity of color
#value - brightness of color

#LAB - lightness and color-opponent dimensions
#L - lightness
#a - green to red
#b - blue to yellow

#BGR - blue, green, red
#OpenCV uses BGR format by default, while most other libraries use RGB. This can lead to confusion when working with images, as the color channels are ordered differently. When you read an image using OpenCV, it will be in BGR format, and if you want to display it using a library that expects RGB, you may need to convert it using cv2.cvtColor() function.
```

```python
import cv2
```

```python
bgr_img = cv2.imread("./data/image.jpeg")
```

```python
gray_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2GRAY)
cv2.imwrite("./data/gray_image.jpeg", gray_img)
```

```python
#convert to HSV
hsv_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2HSV)
cv2.imwrite("./data/hsv_image.jpeg", hsv_img)
```

```python
#convert to LAB
lab_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2LAB)
cv2.imwrite("./data/lab_image.jpeg", lab_img)
```

```python
#convert to rgb
rgb_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2RGB)
cv2.imwrite("./data/rgb_image.jpeg", rgb_img)
#expects bgr format, but we are giving rgb format, so it will save the image in bgr format, which will look like rgb when viewed with a viewer that expects rgb format.
```
